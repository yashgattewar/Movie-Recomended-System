```python
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
movie['genres']=movie['genres'].apply(convert)
```

```python
movie.head()
```

```python
movie['keywords']=movie['keywords'].apply(convert)
```

```python
movie['cast'][0]
```

```python
def convert3(obj):
    L=[]
    counter=0
    for i in ast.literal_eval(obj):
        if counter!=3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
movie['cast']=movie['cast'].apply(convert3)
```

```python
movie.head()
```

```python
def convert4(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
movie['crew'][0]
```

```python
movie['crew']=movie['crew'].apply(convert4)
```

```python
movie.head()
```

```python
movie['overview'][0]
```

```python
movie['overview']=movie['overview'].apply(lambda x:x.split())
```

```python
movie.head()
```

```python
 movie['genres']=movie['genres'].apply(lambda x:[i.replace(" ","") for i in x])
 movie['keywords']=movie['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
 movie['cast']=movie['cast'].apply(lambda x:[i.replace(" ","") for i in x])
 movie['crew']=movie['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

movie.head()


```python
movie['tags']=movie['overview']+movie['genres']+movie['cast']+movie['crew']+movie['keywords']
```

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
final['tags']=final['tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
def recommended(movie):
    movie_index = final[final['title'] == movie].index[0]
    distances = similarity[movie_index]

    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    recommendations = []

    for i in movie_list:
        recommendations.append(final.iloc[i[0]].title)

    return recommendations
```

```python
recomended('Avatar')
```
